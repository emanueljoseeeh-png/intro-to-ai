```python
from collections import deque

def record(trace, current_node, queue_nodes, visited):
    if trace is not None:
        trace.append({
            'current': current_node,
            'queue': list(queue_nodes),
            'visited': list(visited)
        })

def bfs_tree(tree, root, goal, trace=None):
    # Initialize the queue with the starting path containing the root node
    queue = deque()
    queue.append([root])
    visited = []

    while queue:
        # Pop the first path from the queue (FIFO)
        path = queue.popleft()
        node = path[-1]  # the node we are currently standing on
        visited.append(node)

        # Check if we have reached the goal
        if node == goal:
            record(trace, node, [p[-1] for p in queue], visited)
            return path

        # Explore child nodes
        for child in tree.get(node, []):
            queue.append(path + [child])  # new path = old path plus the child

        record(trace, node, [p[-1] for p in queue], visited)

    return None  # Return None if no path to the goal exists

# --- التجربة ---
if __name__ == '__main__':
    tree = {
        'A': ['B', 'C'],
        'B': ['D', 'E'],
        'C': ['F'],
        'D': [],
        'E': [],
        'F': []
    }
    print('المسار:', bfs_tree(tree, 'A', 'E'))
```
