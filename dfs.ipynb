```python
from queue import LifoQueue

def dfs(tree, root, goal):
    stack = LifoQueue()
    stack.put([root])
    visited = []

    while not stack.empty():
        path = stack.get()
        node = path[-1]
        if node not in visited:
            visited.append(node)

            if node == goal:
                return path

            for child in reversed(tree.get(node, [])):
                stack.put(path + [child])

    return None
```
